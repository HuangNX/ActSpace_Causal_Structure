# 根据因果图构建结构因果模型

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from semopy import Model
from semopy import ModelEffects
from semopy.stats import calc_stats
from networkx.drawing import nx_pydot
from sklearn.preprocessing import StandardScaler
```

## 1. data load

变量数据

```python
data = pd.read_csv('../data/act_space_travel_features.csv')
# 删除pid列
data = data.drop('pid', axis=1)

data.head()
```

```python
# 对非int类型的列进行标准化
numeric_cols = data.select_dtypes(include=['float64']).columns
scaler = StandardScaler()
data[numeric_cols] = scaler.fit_transform(data[numeric_cols])

data.head()
```

因果结构

```python
G = nx_pydot.read_dot("supporting_files/graph.dot")
```

```python
# 可视化因果图结构
plt.figure(figsize=(8, 8))  # 设置图形的大小
pos = nx.spring_layout(G, weight=None)  # 布局方式：spring_layout 根据弹簧模型进行布局

# 绘制节点和边
nx.draw(G, pos, with_labels=True, node_size=1000, node_color='skyblue', font_size=10, edge_color='gray')

plt.show()
```

## 2. constuct SEM

```python
# 根据因果图构建semopy模型描述
model_spec = []
for u, v in G.edges():
    # 清理节点名称中的特殊符号（如双引号）
    var_from = u.strip('"')
    var_to = v.strip('"')
    model_spec.append(f"{var_to} ~ {var_from}")  # 转换为回归方程
    
model_desc = '\n'.join(model_spec)
print("Generated SEM model syntax:\n", model_desc)
```

```python
# 创建并拟合模型
model = Model(model_desc)
res = model.fit(data)
print(res)
```

```python
# 获取AGFI (adjusted goodness-of-fit index)
stas_df = calc_stats(model)
stas_df
```

```python
ins = model.inspect()
```

```python
ins.to_csv('results/sem_res1.csv', index=False)
```

## 3. construct random-effect SEM using labeled data

```python
cluster_df = pd.read_csv("../data/act_space_travel_features_cluster.csv")
cluster_df.head()
```

```python
# 对非int类型的列进行标准化
numeric_cols = cluster_df.select_dtypes(include=['float64']).columns
scaler = StandardScaler()
cluster_df[numeric_cols] = scaler.fit_transform(cluster_df[numeric_cols])

cluster_df.head()
```

```python
# 根据label和a_label的组合创建分组
cluster_df['group_str'] = cluster_df.apply(lambda x: f"({int(x['label'])},{int(x['a_label'])})", axis=1)

# 将组合映射为数字标识
unique_groups = cluster_df['group_str'].unique()
group_mapping = {group: idx+1 for idx, group in enumerate(unique_groups)}
cluster_df['group'] = cluster_df['group_str'].map(group_mapping)

# 删除临时列
cluster_df.drop('group_str', axis=1, inplace=True)

cluster_df.head()
```

```python
re_model = ModelEffects(model_desc)
re_res = re_model.fit(cluster_df, group='group')
print(re_res)
```
