```python
import os
import zipfile
from pathlib import Path
```

```python

def compress_folder(folder_path: str,
                    output_zip: str,
                    keep_rel_path = True,
                    compression = zipfile.ZIP_DEFLATED):
    """
    将 `folder_path` 目录递归压缩为 `output_zip`。
    
    Parameters
    ----------
    folder_path : str | Path
        要压缩的目录。
    output_zip : str | Path
        生成的 zip 文件路径（若已存在会被覆盖）。
    keep_rel_path : bool, optional
        - True（默认）：保持原目录层级结构；
        - False：只把文件压到 zip 根目录。
    compression : int, optional
        压缩算法，默认 ZIP_DEFLATED（需要 zlib，Python 默认已集成）。
        可选：ZIP_LZMA, ZIP_BZIP2, ZIP_STORED（不压缩）。
    """
    folder_path = Path(folder_path).resolve()
    output_zip = Path(output_zip).resolve()

    # 确保输出目录存在
    output_zip.parent.mkdir(parents=True, exist_ok=True)

    with zipfile.ZipFile(output_zip, "w", compression=compression) as zf:
        for file in folder_path.rglob("*"):
            if file.is_file():
                if keep_rel_path:
                    arcname = file.relative_to(folder_path)
                else:
                    arcname = file.name
                zf.write(file, arcname)
    print(f"✅ 已将『{folder_path}』压缩为『{output_zip}』")
```

```python
compress_folder("results", "dynamic_causality_results.zip")
```
